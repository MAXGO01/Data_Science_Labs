```python
a = 15
b = 4

print("Сума:", a + b)
print("Різниця:", a - b)
print("Добуток:", a * b)
print("Ділення:", a / b)
print("Ціла частина від ділення:", a // b)
print("Остача від ділення:", a % b)
print("Піднесення до степеня:", a ** b)
```

```python
values = [10, 20, 30, 40, 50]

print("Весь список:", values)
print("Перший елемент:", values[0])
print("Останній елемент:", values[-1])
print("Кількість елементів:", len(values))

values.append(60)
print("Оновлений список:", values)
```
